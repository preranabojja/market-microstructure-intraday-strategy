```python
import sys
import os

# Add project root to Python path
sys.path.append(os.path.abspath(".."))
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/prerana/Documents/market-microstructure-intraday-strategy/data/processed/spy_intraday_backtest_results.csv", parse_dates=['timestamp'])
df = df.set_index('timestamp')

```

```python
#Plot equity curve
plt.figure(figsize=(12,6))
plt.plot(df['cum_return_strategy'], label='Strategy')
plt.plot(df['cum_return_buy_hold'], label='Buy & Hold')
plt.title("Equity Curve")
plt.legend()
plt.show()
```

```python
#Plot drawdown
cum = df['cum_return_strategy']
peak = cum.cummax()
drawdown = (cum - peak) / peak

plt.figure(figsize=(12,4))
plt.plot(drawdown, color='red')
plt.title("Drawdown")
plt.show()
```
