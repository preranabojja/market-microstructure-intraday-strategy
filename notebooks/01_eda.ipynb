```python
import sys
import os

# Add project root to Python path
sys.path.append(os.path.abspath(".."))
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("/Users/prerana/Documents/market-microstructure-intraday-strategy/data/raw/spy_intraday.csv")
df.head()
```

```python
df.columns
```

```python
#Convert to datetime
df['timestamp'] = pd.to_datetime(df['date'])
df = df.set_index('timestamp')
df = df.sort_index()
df.head()
```

```python
#Filter to regular trading hours
df = df.between_time('09:30', '16:00')
df.head()
```

```python
df['close'].plot(figsize=(12,4), title="SPY Intraday Close Prices")
plt.show()
```
